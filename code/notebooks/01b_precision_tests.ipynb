```python
def total(values, chunksize):
    maxsize = len(values)
    indexes = list(range(0, maxsize, chunksize))
    indexes.append(maxsize)
    total = 0
    for i, j in zip(indexes[:-1], indexes[1:]):
        total += sum(values[i:j])
    return total
```
